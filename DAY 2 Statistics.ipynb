# DAY 2 Statistics

```python
#  DAY 2
# Statistics 
import pandas as pd
import numpy as np
print("Day 2 - Statistics and Business Analysis")
print("Libraries loaded successfully.")
```

```python
customers = pd.read_csv("customers_cleaned.csv")
transactions = pd.read_csv("transactions_cleaned.csv")
fraud_alerts = pd.read_csv("fraud_alerts_cleaned.csv")
service_cases = pd.read_csv("service_cases_cleaned.csv")

print("Customers:", customers.shape)
print("Transactions:", transactions.shape)
print("Fraud Alerts:", fraud_alerts.shape)
print("Service Cases:", service_cases.shape)
```

```python
# ============================================
# 1. DESCRIPTIVE STATISTICS
# ============================================

print("========== CUSTOMER STATISTICS ==========")

print("Average Age:", customers["age"].mean())
print("Median Age:", customers["age"].median())
print("Minimum Age:", customers["age"].min())
print("Maximum Age:", customers["age"].max())
print("Standard Deviation of Age:", customers["age"].std())


print("\n========== TRANSACTION STATISTICS ==========")

print("Total Transactions:", transactions["transaction_id"].count())
print("Total Transaction Value:", transactions["amount"].sum())
print("Average Transaction Value:", transactions["amount"].mean())
print("Median Transaction Value:", transactions["amount"].median())
print("Minimum Transaction Value:", transactions["amount"].min())
print("Maximum Transaction Value:", transactions["amount"].max())
print("Standard Deviation:", transactions["amount"].std())


print("\n========== FRAUD ALERT STATISTICS ==========")

print("Total Fraud Alerts:", fraud_alerts["alert_id"].count())
print("Average Anomaly Score:", fraud_alerts["anomaly_score"].mean())
print("Median Anomaly Score:", fraud_alerts["anomaly_score"].median())
print("Minimum Anomaly Score:", fraud_alerts["anomaly_score"].min())
print("Maximum Anomaly Score:", fraud_alerts["anomaly_score"].max())


print("\n========== SERVICE CASE STATISTICS ==========")

print("Total Service Cases:", service_cases["case_id"].count())
print("Average Resolution Time:",
      service_cases["resolution_time_hours"].mean())
print("Median Resolution Time:",
      service_cases["resolution_time_hours"].median())
print("Minimum Resolution Time:",
      service_cases["resolution_time_hours"].min())
print("Maximum Resolution Time:",
      service_cases["resolution_time_hours"].max())
```

```python
# ============================================
# 2. BUSINESS KPIs
# ============================================

total_transactions = len(transactions)

total_transaction_value = transactions["amount"].sum()

average_transaction_value = transactions["amount"].mean()

median_transaction_value = transactions["amount"].median()

successful_transactions = (
    transactions["transaction_status"] == "Success"
).sum()

failed_transactions = (
    transactions["transaction_status"] == "Failed"
).sum()

success_rate = (
    successful_transactions / total_transactions
) * 100

failure_rate = (
    failed_transactions / total_transactions
) * 100

high_value_transactions = (
    transactions["high_amount_flag"] == True
).sum()

international_transactions = (
    transactions["is_international"] == 1
).sum()


print("========== BUSINESS KPIs ==========")

print("Total Transactions:", total_transactions)

print("Total Transaction Value:",
      round(total_transaction_value, 2))

print("Average Transaction Value:",
      round(average_transaction_value, 2))

print("Median Transaction Value:",
      round(median_transaction_value, 2))

print("Successful Transactions:",
      successful_transactions)

print("Failed Transactions:",
      failed_transactions)

print("Success Rate:",
      round(success_rate, 2), "%")

print("Failure Rate:",
      round(failure_rate, 2), "%")

print("High-Value Transactions:",
      high_value_transactions)

print("International Transactions:",
      international_transactions)
```

```python
# ============================================
# 3. CUSTOMER KPIs
# ============================================

print("========== CUSTOMER KPIs ==========")

total_customers = len(customers)

print("Total Customers:", total_customers)

print("\nCustomer Segments:")
print(customers["customer_segment"].value_counts())

print("\nAccount Types:")
print(customers["account_type"].value_counts())

print("\nKYC Status:")
print(customers["kyc_status"].value_counts())

print("\nGender:")
print(customers["gender"].value_counts())
```

```python
# ============================================
# 4. FRAUD KPIs
# ============================================

print("========== FRAUD KPIs ==========")

total_fraud_alerts = len(fraud_alerts)

average_anomaly_score = fraud_alerts["anomaly_score"].mean()

print("Total Fraud Alerts:", total_fraud_alerts)

print("Average Anomaly Score:",
      round(average_anomaly_score, 4))

print("\nRisk Level Distribution:")
print(fraud_alerts["risk_level"].value_counts())

print("\nInvestigation Status:")
print(fraud_alerts["investigation_status"].value_counts())

print("\nAnalyst Action:")
print(fraud_alerts["analyst_action"].value_counts())
```

```python
# ============================================
# 5. SERVICE CASE KPIs
# ============================================

print("========== SERVICE CASE KPIs ==========")

total_cases = len(service_cases)

average_resolution_time = (
    service_cases["resolution_time_hours"].mean()
)

median_resolution_time = (
    service_cases["resolution_time_hours"].median()
)

sla_breached_cases = (
    service_cases["sla_breached"] == True
).sum()

sla_breach_rate = (
    sla_breached_cases / total_cases
) * 100

print("Total Service Cases:", total_cases)

print("Average Resolution Time:",
      round(average_resolution_time, 2), "hours")

print("Median Resolution Time:",
      round(median_resolution_time, 2), "hours")

print("SLA Breached Cases:",
      sla_breached_cases)

print("SLA Breach Rate:",
      round(sla_breach_rate, 2), "%")

print("\nCase Status:")
print(service_cases["status"].value_counts())

print("\nPriority:")
print(service_cases["priority"].value_counts())

print("\nService Channel:")
print(service_cases["channel"].value_counts())
```

```python
# ============================================
# 6. GROUP ANALYSIS - TRANSACTIONS
# ============================================

print("========== TRANSACTION STATUS ANALYSIS ==========")

transaction_status_analysis = (
    transactions
    .groupby("transaction_status")
    .agg(
        transaction_count=("transaction_id", "count"),
        total_value=("amount", "sum"),
        average_value=("amount", "mean")
    )
    .sort_values("transaction_count", ascending=False)
)

transaction_status_analysis
```

```python
# ============================================
# 7. GROUP ANALYSIS - TRANSACTION TYPE
# ============================================

print("========== TRANSACTION TYPE ANALYSIS ==========")

transaction_type_analysis = (
    transactions
    .groupby("transaction_type")
    .agg(
        transaction_count=("transaction_id", "count"),
        total_value=("amount", "sum"),
        average_value=("amount", "mean")
    )
    .sort_values("total_value", ascending=False)
)

transaction_type_analysis
```

```python
# ============================================
# 8. GROUP ANALYSIS - PAYMENT CHANNEL
# ============================================

print("========== PAYMENT CHANNEL ANALYSIS ==========")

payment_channel_analysis = (
    transactions
    .groupby("payment_channel")
    .agg(
        transaction_count=("transaction_id", "count"),
        total_value=("amount", "sum"),
        average_value=("amount", "mean")
    )
    .sort_values("transaction_count", ascending=False)
)

payment_channel_analysis
```

```python
# ============================================
# 9. TREND ANALYSIS - DATE PREPARATION
# ============================================

transactions["transaction_timestamp"] = pd.to_datetime(
    transactions["transaction_timestamp"],
    errors="coerce"
)

transactions["month"] = (
    transactions["transaction_timestamp"]
    .dt.to_period("M")
    .astype(str)
)

transactions["date"] = (
    transactions["transaction_timestamp"]
    .dt.date
)

print("Date conversion completed.")
print("Date range:",
      transactions["transaction_timestamp"].min(),
      "to",
      transactions["transaction_timestamp"].max())
```

```python
# ============================================
# 10. MONTHLY TRANSACTION TREND
# ============================================

monthly_transactions = (
    transactions
    .groupby("month")
    .agg(
        transaction_count=("transaction_id", "count"),
        total_value=("amount", "sum"),
        average_value=("amount", "mean")
    )
    .reset_index()
)

monthly_transactions
```

```python
# ============================================
# 11. MONTHLY SUCCESS / FAILURE TREND
# ============================================

monthly_status = (
    transactions
    .groupby(["month", "transaction_status"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)

monthly_status
```

```python
# ============================================
# 12. DAILY TRANSACTION TREND
# ============================================

daily_transactions = (
    transactions
    .groupby("date")
    .agg(
        transaction_count=("transaction_id", "count"),
        total_value=("amount", "sum"),
        average_value=("amount", "mean")
    )
    .reset_index()
)

print("DAILY TRANSACTION TREND")
print(daily_transactions.head(10))

print("\nTotal Days:", len(daily_transactions))
print("Highest Transaction Day:")
print(
    daily_transactions.loc[
        daily_transactions["transaction_count"].idxmax()
    ]
)

print("\nHighest Transaction Value Day:")
print(
    daily_transactions.loc[
        daily_transactions["total_value"].idxmax()
    ]
)
```

```python
# ============================================
# 13. WEEKLY TRANSACTION TREND
# ============================================

transactions["week"] = (
    transactions["transaction_timestamp"]
    .dt.to_period("W")
    .astype(str)
)

weekly_transactions = (
    transactions
    .groupby("week")
    .agg(
        transaction_count=("transaction_id", "count"),
        total_value=("amount", "sum"),
        average_value=("amount", "mean")
    )
    .reset_index()
)

print("WEEKLY TRANSACTION TREND")
print(weekly_transactions.head(10))

print("\nTotal Weeks:", len(weekly_transactions))
```
